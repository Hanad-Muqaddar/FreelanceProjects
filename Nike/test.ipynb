```python
import requests
result = requests.get("https://www.nike.com/")
```

```python
from scrapy import Selector
response = Selector(result)
```

```python

links = response.css("#DesktopMenu-0-4-0 .pre-text-color-secondary ::attr(href)").getall()
print(links)
```
